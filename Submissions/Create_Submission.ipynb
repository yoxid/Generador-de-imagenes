# Create random images

```python
import numpy as np
from PIL import Image
import os

n_images = 200
width = 64
height = 64
channels = 3

out_folder = "random_imgs"
os.makedirs(out_folder, exist_ok=True)

random_data = np.random.randint(0, 256, (n_images, height, width, channels), dtype=np.uint8)

for i in range(n_images):
    img = random_data[i]
    img = Image.fromarray(img, 'RGB')
    nombre_archivo = f"{out_folder}/img_{i+1:03d}.png"
    img.save(nombre_archivo)
```

# Evaluate created images

```python
import glob
import cv2
import numpy as np
import pandas as pd



#sufix = "ae"
#sufix = "ae_random"
#sufix = "vae"
sufix = "vae_random"


out_folder = "imgs_" + sufix


# Load images in b&w 64x64 resolution
imgs_arr = []
for img_path in glob.glob(out_folder + '/*.png'):
    img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
    img = cv2.resize(img, (64, 64), interpolation=cv2.INTER_AREA)
    imgs_arr.append(img.flatten())
df = pd.DataFrame(np.array(imgs_arr))


df.insert(0, "ID", range(1, len(df) + 1))
# Create .csv
df.to_csv(f'submission_{sufix}.csv', index=False)
```
